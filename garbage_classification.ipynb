# GARBAGE CLASSIFICATION MODEL
### Logistic Regression Classification
This notebook classifies waste into **Biodegradable** and **Non-Biodegradable** using the supplied CSV dataset.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, confusion_matrix, classification_report,
    precision_score, recall_score, f1_score
)
```

## 2. Load Dataset

```python
df = pd.read_csv("garbage_classification_dataset_5000.csv")
print(df.head())
```

## 3. Understand the Dataset

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nClass distribution:")
print(df["Waste_Type"].value_counts())
```

## 4. Data Visualization

```python
df["Waste_Type"].value_counts().plot(
    kind="bar", rot=0, title="Garbage Class Distribution"
)
plt.xlabel("Waste Type")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
for waste_type in df["Waste_Type"].unique():
    subset = df[df["Waste_Type"] == waste_type]
    plt.scatter(subset["Weight_g"], subset["Moisture_Percent"],
                label=waste_type, alpha=0.5)
plt.xlabel("Weight (g)")
plt.ylabel("Moisture (%)")
plt.title("Weight vs Moisture")
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Select Features (X) and Target (y)

```python
x = df[["Weight_g", "Moisture_Percent", "Size_cm3"]]
y = df["Waste_Type"]

print("Features:")
display(x.head())
print("Target:")
display(y.head())
```

## 6. Split Dataset into Training and Testing

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42, stratify=y
)
print("Training records:", len(x_train))
print("Testing records:", len(x_test))
```

## 7. Train Logistic Regression Model

```python
model = Pipeline([
    ("scaler", StandardScaler()),
    ("logistic_regression", LogisticRegression(max_iter=1000))
])
model.fit(x_train, y_train)
print("Model training completed!")
```

## 8. Make Predictions

```python
y_pred = model.predict(x_test)
comparison = pd.DataFrame({"Actual": y_test, "Predicted": y_pred})
display(comparison.head(20))
```

## 9. Evaluate the Model

```python
cm = confusion_matrix(
    y_test, y_pred,
    labels=["Biodegradable", "Non-Biodegradable"]
)
print("Confusion Matrix (rows = actual, columns = predicted):")
print(cm)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred,
                            pos_label="Biodegradable", zero_division=0)
recall = recall_score(y_test, y_pred,
                      pos_label="Biodegradable", zero_division=0)
f1 = f1_score(y_test, y_pred,
              pos_label="Biodegradable", zero_division=0)

print("\nAccuracy:", round(accuracy, 3))
print("Precision:", round(precision, 3))
print("Recall:", round(recall, 3))
print("F1 Score:", round(f1, 3))

print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

## 10. Predict New Garbage

```python
new_garbage = pd.DataFrame({
    "Weight_g": [80],
    "Moisture_Percent": [65],
    "Size_cm3": [400]
})

predicted_class = model.predict(new_garbage)[0]
predicted_probabilities = model.predict_proba(new_garbage)[0]

print("New garbage prediction:", predicted_class)
print("Class order:", model.classes_)
print("Class probabilities:", predicted_probabilities)
```

> Note: This is a synthetic practice dataset based on numeric waste features, not an image-based garbage recognition model.
